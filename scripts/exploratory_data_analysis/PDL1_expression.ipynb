```python
import pandas as pd
import matplotlib.pyplot as plt
from tex import tex_gen_file

# import matplotlib as mpl
mpl.rcParams.update(mpl.rcParamsDefault)
```

```python
y_broad = pd.read_csv("../../data/Y_PDL1expression_Broad.csv")["PDL1 expression"]
y_broad.info()
```

```python
y_broad.describe()
```

```python

cutoff = 0.5

ax = y_broad[y_broad < cutoff].hist(bins=100, density=True, grid=False)
ax.set_xlim(y_broad.min(), cutoff)
ax.set_ylim(bottom=0)

quantiles = [0.5, 0.6, 0.7, 0.8, 0.9]
for q in quantiles:
    ax.axvline(y_broad.quantile(q), color="pink", alpha=0.5, linestyle=":", label=f"{int(q*100)}% quantile")
ax.legend()

ax.set_ylabel("Frequency")
ax.set_xlabel("PD-L1 expression")
ax.set_title("Histogram of PD-L1 expression")

plt.savefig(tex_gen_file("data/pdl1_expression_broad.svg", root="../../"))
plt.show()
```
